```python
for numero in range(-30, 51):
    if numero < 0:
        categoria = "Muy frío"
    elif numero <= 24:
        categoria = "Templado"
    else:
        categoria = "Caluroso"

    print(f"{numero} {categoria}")
```

```python
for numero in range(1, 101):
    if numero % 10 == 0:
        print(f"{numero} Bip Bip")
    elif numero % 5 == 0:
        print(f"{numero} Bip")
```

```python
# Entrada de datos
puntaje1 = int(input("Ingresa primer puntaje: "))
puntaje2 = int(input("Ingresa segundo puntaje: "))

# Cálculo
suma = puntaje1 + puntaje2

# Evaluación de nivel
if suma < 50:
    nivel = "Nivel principiante"
elif suma < 80:
    nivel = "Nivel intermedio"
else:
    nivel = "Nivel pro"

# Salida
print(f"\nResultado: {nivel}")
```

```python
edad = int(input("Ingresa tu edad: "))
le_gusta_musica = input("¿Te gusta la música? (si/no): ").strip().lower() == "si"
le_gusta_bailar = input("¿Te gusta bailar? (si/no): ").strip().lower() == "si"

if edad >= 18 and le_gusta_musica and le_gusta_bailar:
    print("Eres una estrella de la fiesta")
elif le_gusta_musica and not le_gusta_bailar:
    print("Te gusta la música, pero no tanto bailar")
else:
    print("Tienes tu propio estilo")
```
